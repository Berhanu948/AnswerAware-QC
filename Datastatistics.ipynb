```python
#Data Statistics
import pandas as pd
from sklearn.model_selection import train_test_split
# ==============================
# 1. Load the dataset
# ==============================
df=pd.read_csv("AmHQA.csv")
# Check the column names
print("Columns:", df.columns.tolist())
# ==============================
# 2. Standardize Question Type
# ==============================
# Change this if your column has a different name
df["Question Type"] = df["Category"].astype(str).str.strip().str.title()

# Display available question types
print("\nQuestion Types:")
print(df["Question Type"].value_counts())

# ==============================
# 3. Create stratified splits
#    80% Training, 10% Validation, 10% Test
# ==============================

# First split: 80% training, 20% temporary
train_df, temp_df = train_test_split(
    df, test_size=0.20, stratify=df["Question Type"],random_state=42)

# Second split: divide remaining 20% equally
validation_df, test_df = train_test_split(temp_df,test_size=0.50,stratify=temp_df["Question Type"],random_state=42
)

# ==============================
# 4. Calculate counts
# ==============================
question_types = [
    "Factoid",
    "List",
    "Definition",
    "Description"
]

rows = []

for qtype in question_types:
    original = (df["Question Type"] == qtype).sum()
    training = (train_df["Question Type"] == qtype).sum()
    validation = (validation_df["Question Type"] == qtype).sum()
    test = (test_df["Question Type"] == qtype).sum()

    rows.append([
        qtype,
        original,
        training,
        validation,
        test
    ])

# Add total row
rows.append([
    "Total",
    len(df),
    len(train_df),
    len(validation_df),
    len(test_df)
])

# ==============================
# 5. Create summary table
# ==============================
summary = pd.DataFrame(
    rows,
    columns=[
        "Question Type",
        "Original Dataset",
        "Training",
        "Validation",
        "Test"
    ]
)

print("\nDataset Distribution:")
print(summary.to_string(index=False))

# ==============================
# 6. Optional: save the table
# ==============================
summary.to_csv(
    "amqc_dataset_distribution.csv",
    index=False
)
```
